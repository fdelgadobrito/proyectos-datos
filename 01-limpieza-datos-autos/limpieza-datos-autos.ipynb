## **3.4. Valores nulos (missing)**

Identificacion de datos nulos.

```python
#Datos nulos
df.isna().sum()
```

La base de datos presenta valores nulos en distintas proporciones, algunas columnas con pocos casos (como Fecha_Venta o Marca) y otras con miles de registros afectados (como Region, Antigüedad o Vigencia_Garantia). Antes de eliminar todas las filas con nulos, vamos analizarlas si hay hay variables que se puedan deducir en otras, y asi se eliminarán las filas con pocos nulo. En este sentido, se calcularán directamente las columnas derivables de otras (como Antigüedad o Region); y se imputarán mediante agrupación (por Marca, Modelo, Año_Fabricacion) las columnas como Kilometraje, Precio_Venta, Transmision o Combustible, usando la mediana o moda del grupo correspondiente.

Eliminación de registros con menor cantidad de nulos, ademas de "precio venta", que es la variable principal de análisis:

```python
# Eliminar filas con datos nulos
df = df.dropna(subset=['Fecha_Venta','Ciudad','Precio_Venta'])
```

```python
# Eliminar filas con datos nulos
df = df.dropna(subset=['Tipo','Modelo','Color','Año_Fabricacion','Transmision'])
```


Imputación de datos faltantes de las variables con mayor cantidad de nulos.

```python
'''
Imputacion por moda basado en la variable marca y modelo, para variables categóricas:

'''

# Calculo de la moda
df['Combustible'] = df['Combustible'].fillna(
    df.groupby(['Marca', 'Modelo'])['Combustible']
      .transform(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
)
# Verificar
print(df[['Combustible']].isna().sum())
```

Lo mismo anterior, pero basado en mediana y otras variables base para ser mas preciso.

```python
# Imputar Kilometraje con mediana agrupada (Modelo + Año_Fabricacion)
df['Kilometraje'] = df['Kilometraje'].fillna(
    df.groupby(['Modelo', 'Año_Fabricacion'])['Kilometraje'].transform('median')
)

# Si quedan nulos porque el grupo específico no tenía datos, usar un grupo más amplio
df['Kilometraje'] = df['Kilometraje'].fillna(
    df.groupby(['Modelo'])['Kilometraje'].transform('median')
)

# Último recurso: mediana general de toda la columna
df['Kilometraje'] = df['Kilometraje'].fillna(df['Kilometraje'].median())

# Verificar
print(df['Kilometraje'].isna().sum())
```

```python
#Metodo de pago a traves de moda
df['Metodo_Pago'] = df['Metodo_Pago'].fillna(df['Metodo_Pago'].mode()[0])

# Verificar
print(df['Metodo_Pago'].isna().sum())
```

Para el Tipo_Garantia vamos asumir que la falta de datos significa "Sin Garantía", asi como tambien, para el Equipamiento, es "Básico"

```python
# Tipo_Garantia: el nulo probablemente significa que el vehículo no tiene garantía
df['Tipo_Garantia'] = df['Tipo_Garantia'].fillna('Sin Garantia')

# Equipamiento: el nulo probablemente significa que no tiene equipamiento adicional
df['Equipamiento'] = df['Equipamiento'].fillna('Básico')

# Verificar
print(df[['Tipo_Garantia', 'Equipamiento']].isna().sum())
```

```python
# Imputación de datos categóricos basados en otras variables:

# Paso 1: Función para asignar cada región por ciudad
def asignar_region(ciudad):
    mapeo_ciudad_region = {
        'Santiago': 'Metropolitana',
        'Las Condes': 'Metropolitana',
        'Maipú': 'Metropolitana',
        'La Dehesa': 'Metropolitana',
        'Vitacura': 'Metropolitana',
        'Providencia': 'Metropolitana',
        'Valparaíso': 'Valparaíso',
        'Viña del Mar': 'Valparaíso',
        'Concepción': 'Biobío',
        'Los Ángeles': 'Biobío',
        'Rancagua': "O'Higgins",
        'San Fernando': "O'Higgins",
        'Talca': 'Maule',
        'Curicó': 'Maule',
        'Linares': 'Maule',
        'Chillán': 'Ñuble',
        'Quillón': 'Ñuble',
        'Temuco': 'Araucanía',
        'Villarrica': 'Araucanía',
        'Valdivia': 'Los Ríos',
        'La Unión': 'Los Ríos',
        'Osorno': 'Los Lagos',
        'Puerto Montt': 'Los Lagos',
        'Puerto Varas': 'Los Lagos',
        'Coyhaique': 'Aysén',
        'Punta Arenas': 'Magallanes',
        'Arica': 'Arica y Parinacota',
        'Iquique': 'Tarapacá',
        'Pozo Almonte': 'Tarapacá',
        'Antofagasta': 'Antofagasta',
        'Calama': 'Antofagasta',
        'Copiapó': 'Atacama',
        'Vallenar': 'Atacama',
        'La Serena': 'Coquimbo',
        'Coquimbo': 'Coquimbo',
    }
    return mapeo_ciudad_region.get(ciudad, None)

# Paso 2: Aplicar la función SOLO donde Region está vacío,
# para no sobrescribir los valores que ya eran correctos
df['Region'] = df['Region'].fillna(df['Ciudad'].apply(asignar_region))
```

Vamos a impugnar los años de antiguedad basado en la variable año de fabricacion y fecha de venta.

vencimiento de garantia, debe ser igual a FECHA DE VENTA+ TIPO DE GARANTIA, indicando cuando se vencio la garantia en su fecha exacta

```python
# Vencimiento_Garantia, los datos los pasamos a diccionario para darle valor numerico
duracion_garantia = {
    '6 meses': 6,
    '1 año': 12,
    '2 años extendida': 24,
    'Sin Garantía': 0   # duración 0 = la fecha de vencimiento es la misma que la de venta
}

df['Vencimiento_Garantia'] = df.apply(
    lambda row: row['Fecha_Venta'] + pd.DateOffset(months=duracion_garantia.get(row['Tipo_Garantia'], 0)),
    axis=1
)

#  Vigencia_Garantia// aca modificamos la lectura de la ultima variable para que sea mas fácil su tratamiento entre Vencida o Vigente
hoy = pd.Timestamp.today()

def calcular_vigencia(row):
    if row['Tipo_Garantia'] == 'Sin Garantía':
        return 'Vencida'  # Sin garantía siempre se considera vencida
    return 'Vigente' if row['Vencimiento_Garantia'] >= hoy else 'Vencida'

df['Vigencia_Garantia'] = df.apply(calcular_vigencia, axis=1)

# Verificar
print(df['Vencimiento_Garantia'].isna().sum())
print(df['Vigencia_Garantia'].value_counts())
df.sample(10)
```

```python
# Importar librerías de visualización
import matplotlib.pyplot as plt

# Función personalizada para graficar boxplot de varias columnas a la vez
def graficar_boxplot(df, columnas):
    """
    Genera boxplots lado a lado para columnas especificadas de un DataFrame.

    Parámetros:
    - df: DataFrame de pandas
    - columnas: lista de nombres de columnas numéricas a graficar
    """
    subset = df[columnas]
    fig, axes = plt.subplots(1, len(columnas), figsize=(6 * len(columnas), 4))

    if len(columnas) == 1:
        axes = [axes]

    for i, column in enumerate(columnas):
        ax = axes[i]
        ax.boxplot(subset[column])
        ax.set_title(f"Boxplot de {column}")
        ax.set_ylabel("Valor")
        ax.get_xaxis().set_visible(False)
        # Evita notación científica en eje y
        ax.ticklabel_format(style='plain', axis='y')

    plt.tight_layout()
    plt.show()
```

```python
# Usar la función personalizada para graficar boxplot
graficar_boxplot(df, ["Precio_Venta", "Kilometraje","Antigüedad"])
```

Se realiza una inspeccion visual de las variables numéricas y se encuentran datos atípicos en Impuestos y Monto.

Se puede realizar el tratamiento de datos atípicos mediante el método Z-score. Este permite describir un punto de datos en relación con su desviación media y estándar de un conjunto de datos, así realizando un corte usualmente de una puntuación de Z = 3 (es decir, 3 desviaciones desde la media) para considerar fuera de ese rango un valor atípico. Entre mayor sea el numero de desviaciones estandar en el umbral de corte, menos valores se considerarán atípicos y viceversa

```python
# Importar librería de estadísticas
from scipy import stats

# Se define la cantidad de desviaciones estandar a considerar para el umbral de corte
corte = 3
```

```python
# Calcula el Z-Score para 'Precio_Venta'
z_precioventa = stats.zscore(df['Precio_Venta'])
outliers_precioventa = (abs(z_precioventa) > corte)
# Imprime el recuento de valores atípicos en 'Precio_Venta'
print(f"Recuento de valores atípicos en Precio_Venta: {outliers_precioventa.sum()}")
```

```python
# Elimina las filas con valores atípicos
df = df[~outliers_precioventa]
```

```python
# Calcula el Z-Score para 'Antigüedad'
z_antigüedad = stats.zscore(df['Antigüedad'])
outliers_antigüedad = (abs(z_antigüedad) > corte)
# Imprime el recuento de valores atípicos en 'Antigüedad'
print(f"Recuento de valores atípicos en Antigüedad: {outliers_antigüedad.sum()}")
```

```python
# Elimina las filas con valores atípicos
df = df[~outliers_antigüedad]
```

Se corrobora la eliminación de datos atípicos:

```python
# Usar la función personalizada para graficar boxplot
graficar_boxplot(df, ["Precio_Venta", "Antigüedad"])
```

EVIDENCIAMOS QUE PRECIO_VENTA ya se encuentra con numeros acordes y coherentes, pero en ANTIGUEDAD, nos encontramos con valores minimos -4 años que no tiene coherencia, los cuales los vamos a identificar y dependiendo la cantidad de filas lo vamos a eliminar o modificar si es del caso.

```python
#Montos bajo 0 años
len(df[df['Antigüedad'] <= 0])
```

Los valores de ANTIGUEDAD menores o iguales a 0 años no son lógicos. Procederemos a convertirlos en NAN y luego a imputarlos utilizando el promedio de ANTIGUEDAD agrupado por KILOMETRAJE

```python
# Reemplazar valores ilógicos (<= 0) en 'Antigüedad' con NaN
df.loc[df['Antigüedad'] <= 0, 'Antigüedad'] = np.nan

# Imputar NaN en 'Antigüedad' usando la mediana agrupada por 'Kilometraje'
df['Antigüedad'] = df.groupby('Kilometraje')['Antigüedad'].transform(lambda x: x.fillna(x.mean()))

# Si todavía quedan NaN (por ejemplo, si un grupo de Kilometraje solo tenía valores <=0),
# rellenar con la media general de 'Antigüedad'.
df['Antigüedad'] = df['Antigüedad'].fillna(df['Antigüedad'].mean())

# Verificar si aún quedan valores nulos en 'Antigüedad'
print(f"Valores nulos restantes en 'Antigüedad': {df['Antigüedad'].isna().sum()}")

# Mostrar el resumen estadístico actualizado para verificar la columna 'Antigüedad'
df[['Precio_Venta', 'Antigüedad']].describe()
```

```python
# Resumen estadístico de variables numéricas
df[['Precio_Venta', 'Antigüedad', 'Kilometraje']].describe()
```

Se arreglaron los valores de ATIGUEDAD, pero se evidencia que en KILOMETRAJE tambien hay valores no coherentes.... se hará el mismo procedimiento anterior. pero al revés.

```python
len(df[df['Kilometraje'] <= 0])
```

Se evidencia que existen valores no coherentes de Kilometraje, los cuales pasaremos a NaN y luego imputaremos a través de la mediana.

```python
# Reemplazar valores ilógicos (<= 0) en 'Kilometraje' con NaN
df.loc[df['Kilometraje'] <= 0, 'Kilometraje'] = np.nan

# Imputar NaN en 'Kilometraje' usando la mediana agrupada por 'Modelo' y 'Año_Fabricacion'
df['Kilometraje'] = df.groupby(['Modelo', 'Año_Fabricacion'])['Kilometraje'].transform(lambda x: x.fillna(x.median()))

# Si quedan nulos porque el grupo específico no tenía datos, usar un grupo más amplio (por 'Modelo')
df['Kilometraje'] = df.groupby('Modelo')['Kilometraje'].transform(lambda x: x.fillna(x.median()))

# Último recurso: mediana general de toda la columna
df['Kilometraje'] = df['Kilometraje'].fillna(df['Kilometraje'].median())

# Verificar si aún quedan valores nulos en 'Kilometraje'
print(f"Valores nulos restantes en 'Kilometraje': {df['Kilometraje'].isna().sum()}")

# Mostrar el resumen estadístico actualizado para verificar la columna 'Kilometraje'
df[['Precio_Venta', 'Antigüedad', 'Kilometraje']].describe()
```
